```python
#Load and Explore the Dataset#
import numpy as np
import pandas as pd

# ============================================
# Load California Housing Dataset
# ============================================

try:
    url = "https://raw.githubusercontent.com/ageron/handson-ml2/master/datasets/housing/housing.csv"
    df = pd.read_csv(url)
    print("Dataset loaded successfully from URL!")

except Exception as e:
    print(f"URL download failed: {e}")
    print("Generating synthetic dataset...")

    np.random.seed(42)
    n_samples = 5000

    df = pd.DataFrame({
        "longitude": np.random.uniform(-124.5, -114.5, n_samples),
        "latitude": np.random.uniform(32.5, 42.0, n_samples),
        "housing_median_age": np.random.randint(1, 52, n_samples),
        "total_rooms": np.random.normal(2000, 1000, n_samples).astype(int),
        "total_bedrooms": np.random.normal(500, 250, n_samples).astype(int),
        "population": np.random.normal(1200, 800, n_samples).astype(int),
        "households": np.random.normal(450, 250, n_samples).astype(int),
        "median_income": np.random.uniform(0.5, 15, n_samples),
        "median_house_value": np.random.normal(200000, 80000, n_samples).astype(int),
        "ocean_proximity": np.random.choice(
            ["NEAR BAY", "<1H OCEAN", "INLAND", "NEAR OCEAN", "ISLAND"],
            n_samples,
            p=[0.30, 0.30, 0.30, 0.09, 0.01]
        )
    })

    # Create a relationship between features and house price
    df["median_house_value"] = (
        df["median_income"] * 40000 +
        df["housing_median_age"] * 1000 +
        df["total_rooms"] * 20 +
        np.random.normal(0, 30000, n_samples)
    ).astype(int)

    # Limit house values to realistic values
    df["median_house_value"] = df["median_house_value"].clip(
        lower=50000,
        upper=500001
    )

    print("Synthetic dataset created successfully!")
```

```python
#Part 3: Exploratory Data Analysis (EDA)#
#Distribution of the Target Variable#
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# --------------------------
# Histogram
# --------------------------

axes[0].hist(
df["median_house_value"],
bins=50,
edgecolor="black",
alpha=0.7
)

axes[0].set_title("Distribution of House Values")
axes[0].set_xlabel("House Value ($)")
axes[0].set_ylabel("Frequency")
axes[0].axvline(
df["median_house_value"].median(),
color="red",
linestyle="--",
label=f"Median: ${df['median_house_value'].median():,.0f}"
)

axes[0].axvline(
df["median_house_value"].mean(),
color="green",
linestyle="--",
label=f"Mean: ${df['median_house_value'].mean():,.0f}"
)

axes[0].legend()

# --------------------------
# Box Plot
# --------------------------

axes[1].boxplot(df["median_house_value"])

axes[1].set_title("Box Plot of House Values")
axes[1].set_ylabel("House Value ($)")

# --------------------------
# Log Transformation
# --------------------------

axes[2].hist(
np.log1p(df["median_house_value"]),
bins=50,
edgecolor="black",
alpha=0.7
)

axes[2].set_title("Log-Transformed Distribution")
axes[2].set_xlabel("Log(House Value)")
axes[2].set_ylabel("Frequency")

plt.tight_layout()

plt.show()
```

Interpretation
The histogram shows how house prices are distributed.
The box plot helps identify possible outliers.
The log-transformed histogram shows whether the data becomes more normally
distributed after applying a logarithmic transformation.

```python
#Correlation Analysis#
numeric_columns = df.select_dtypes(include=[np.number]).columns
correlations = (
df[numeric_columns]
.corr()["median_house_value"]
.sort_values(ascending=False)
)
```

```python
print("Top Correlations with House Price")
print(correlations)
```

```python
#Visualize Correlations#
plt.figure(figsize=(10, 6))

correlations.drop("median_house_value").plot(kind="bar")

plt.title("Feature Correlation with House Price")

plt.xlabel("Features")

plt.ylabel("Correlation Coefficient")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

This graph helps identify which features have the strongest relationship with house prices.

```python
#Scatter Plots#
fig, axes = plt.subplots(2, 3, figsize=(15, 10))

features = [
    "median_income",
    "housing_median_age",
    "total_rooms",
    "population",
    "households",
    "latitude"
]

for index, feature in enumerate(features):
    row = index // 3
    column = index % 3

    # Scatter plot
    axes[row, column].scatter(
        df[feature],
        df["median_house_value"],
        alpha=0.3,
        s=10
    )

    axes[row, column].set_title(f"House Value vs {feature}")
    axes[row, column].set_xlabel(feature)
    axes[row, column].set_ylabel("House Value ($)")

    # Add linear trend line
    z = np.polyfit(df[feature], df["median_house_value"], 1)
    p = np.poly1d(z)
    
    # Generate clean x range for the trendline
    x_range = np.linspace(df[feature].min(), df[feature].max(), 100)
    
    axes[row, column].plot(
        x_range,
        p(x_range),
        "r--",
        linewidth=2
    )

plt.tight_layout()
plt.show()
```

Interpretation
Observe each scatter plot carefully and answer the following questions:
1. Does the feature show a positive or negative relationship with house price?
 Direction of Relationship (Positive or Negative)
median_income: Strong positive relationship. As income increases, median house value rises steadily.

housing_median_age: Slightly positive relationship. Older neighborhoods tend to have moderately higher values, though the points are widely spread.

total_rooms: Weak positive relationship. More rooms generally correlate with larger, higher-value homes.

population & households: Weak positive to near-flat relationship. Overall counts correlate slightly with house values, but density doesn't directly dictate high per-house prices.

latitude: Weak negative relationship. Higher latitudes (moving north in California toward interior rural areas) generally correlate with lower house prices compared to coastal southern/central areas.

2. Is the relationship strong or weak? 2. Strength of Relationship
Strong: median_income (the data points tightly cluster along the upward trend line).

Weak / Moderate: housing_median_age, total_rooms, latitude, population, and households (high variance and wide dispersion around the trend line).

3. Are there any outliers?3. Outliers
Upper Cap / Price Ceiling: A noticeable horizontal line of data points appears at the top boundary ($500,001). These are capped values where houses priced above $500,000 were clipped in the dataset collection process.

Feature Extremes: total_rooms, population, and households show extreme horizontal outliers to the right (a few large developments or block groups with massive total counts).

4. Which feature appears to have the greatest influence on house prices?4. Greatest Influence Feature
median_income clearly has the strongest influence on house prices. It exhibits the highest linear correlation, the narrowest scatter band along its positive trend line, and serves as the single best predictor for median_house_value.
